```python
import matplotlib.pyplot as plt
plt.rcParams.update({'font.size': 20})

import pandas as pd
import numpy as np

import datetime
import os

%matplotlib inline

from graphics_utils.plot_utils import *
def tensor_str_to_float(x):
    try:
        return float(x[9:15])
    except Exception as e:
        print(e)
        return np.nan
def unlist(x):
    return float(x.replace('[','').replace(']',''))

from PIL import Image
```

```python
verbose = 0
```

```python
def get_mean_interpolated_stats_df_wrapper(log_path, **kwargs):
    new_x, new_y, runs, interp_funcs = get_mean_interpolated_stats_df(log_path, **kwargs)
    return new_x, new_y, runs
```

```python
def draw_meanIntep_with_colored_runs_wrapper(new_x, new_y, runs, **kwargs):
    ax = draw_meanIntep_with_colored_runs(
        new_y, runs, new_x, **kwargs
    )
    return ax
```

```python
def draw_mean_with_std_wrapper(new_x, new_y, runs, **kwargs):
    ax = draw_mean_with_std(new_y, runs, new_x, **kwargs)
    return ax
```
